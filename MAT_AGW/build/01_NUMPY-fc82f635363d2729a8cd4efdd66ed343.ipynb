Der **ndarray** (nachfolgend einfach **Array** genannt) ist die zentrale Datenstruktur der NumPy-Bibliothek und kann ein-, zwei- oder auch höherdimensional sein. Da der Array viele Ähnlichkeiten mit der Liste aufweist, können wir bzgl. Erstellung und Indizierung auf das vorhandene Wissen aufbauen.

````python
a = np.array([-1.23, 0, 2.5])  # manuelle Erstellung eines Arrays
print(a)
type(a)
````

### 2.1 Erstellung

Für die Erstellung von 1D-Arrays wird meistens eine der folgenden Funktionen verwendet:

- `np.array()` - für die manuelle Erzeugung (oder Konversion von einem anderen Datentyp)
- `np.arange()` - analog zur Funktion `range()`, aber nicht auf Ganzzahlen beschränkt
- `np.linspace()` - wie `np.arange()`, aber mit Sampleanzahl anstelle der Schrittweite
- `np.logspace()` - wie `np.linspace()`, aber mit logarithmischer Teilung

````python
a = np.array([-1, 0, 1, 2])  # manuell, ähnlich zur Erstellung einer Liste
print(a)
a = np.arange(0, 1, step=0.2) # (Startwert, Endwert (nicht inkl.), Schrittweite)
print(a)
a = np.linspace(0, 5, num=9)  # (Startwert, Endwert (inkl.), Sampleanzahl)
print(a)
a = np.linspace(0, 5, num=9, endpoint=False)  # wie oben, aber diesmal Endwert nicht inkl.
print(a)
a = np.logspace(-2, 3, num=6, base=10.0)  # logarithmische Teilung
print(a)
````

Besonders erwähnenswert im Zusammenhang mit der Indizierung ist die **boolesche Indizierung** (*boolean indexing*), die sehr praktisch ist, um einen Array anhand seiner Elemente (Werte) zu "filtern", d. h. nur bestimmte Elemente zu identifizieren.

````python
print(a[a>2])  # alle Elemente aus a, die grösser als 2 sind
print(a[(a>1)&(a<2)])  # alle Elemente aus a, die grösser als 1 und kleiner als 2 sind
a[(a>=2.5)|(a<=0.5)] = 0  # alle Elemente aus a, die grösser/gleich 2.5 oder kleiner/gleich 0.5 sind auf 0 setzen
print(a)
````

Dies wird boolesche Indizierung genannt, weil in den `[]`-Klammern dabei ein boolescher Array steht, der für jedes Element entweder *True* oder *False* ist.

````python
print(a>1)  # Vergleiche führen zu booleschem Array
````

````python
print(a[a>1])  # Indizierung mit booleschem Array (siehe auch Bsp. oben)
````

### 3.3 Editieren/Kopieren

Elemente in Arrays können **verändert** werden, wie wir dies von Listen kennen. Wie bei Listen muss beachtet werden, dass eine Zuweisung eines Arrays `a` zur Variable `b`, also `b = a`, keine Kopie aller Arrayelemente vornimmt, sondern lediglich die Referenz auf den Array kopiert. Der Array `b` "enthält" dann dieselben Elemente wie `a` und wenn ein Element in `b` verändert wird, so zeigt sich diese Änderung auch in `a`.

````python
a = np.array([1, 2, 3])
b = a  # keine Kopie, nur Zuweisung (Referenz)
b[1] = -99  # auch a wird verändert
print('a: ', a)
print('b: ', b)
````

Eine wirkliche (eigenständige) Kopie kann mit der Methode `.copy()` erzeugt werden.

````python
a = np.array([1, 2, 3])
b = a.copy()  # Kopie
b[1] = -99  # a ist nicht betroffen
print('a: ', a)
print('b: ', b)
````

Ein grosser Vorteil von NumPy liegt in der Einfachheit, mit der mathematische Berechnungen mit Arrays durchgeführt werden können. Am Beispiel der folgenden Formel zur Berechnung des quadratischen Mittelwertes (Effektivwert) $RMS$ (*root mean square*) von $N$ Zahlen wollen wir dies illustrieren. 

$$RMS = \sqrt{ \frac{1}{N} \sum_{i=0}^{N} x_i^2 }$$

☝🏼 Würde man mit einer Liste arbeiten, könnte die Berechnung nicht so direkt umgesetzt werden und es wäre eine Schleife oder z. B. die `map`-Funktion nötig.

````python
N = 100
x = np.random.standard_normal(N)  # 100 standardnormalverteilte Zufallszahlen

RMS = np.sqrt(1/N * np.sum(np.square(x)))  # Die obige Formel mit NumPy
print(RMS)
````

## A. Anhang

Hier folgen weiterführende Informationen zum NumPy-Array, die wir allerdings im Unterricht nicht benötigen und daher als optional zu betrachten sind.

### A.1 Elementdatentyp

Da NumPy für maximale Performance entwickelt wurde, unterscheidet sich der Array dahingehend von der Liste, dass sämtliche Elemente im Array denselben Datentyp aufweisen. Dieser Elementdatentyp kann optional bei der Erstellung zugewiesen werden und es steht eine Reihe von spezifischen Datentypen zur Verfügung, wobei die Zahl jeweils für die entsprechende Bitgrösse steht:

- `np.uint8/16/32/64` - vorzeichenlose Ganzzahlen (*unsigned integer*)
- `np.int8/16/32/64` - vorzeichenbehaftete Ganzzahlen (*signed integer*)
- `np.float16/32/64/96/128` - Gleitkommazahlen (*floating point*)
- `np.complex64/128/192/256` - komplexe Zahlen (*complex*)

Dies ermöglicht es, für eine bestimmte Anwendung den optimalen (in Bezug auf den Speicherplatz kleinstmöglichen) Datentyp zu wählen.

☝🏼 Der Elementdatentyp `uint16` beinhaltet z. B. nur vorzeichenlose (*u* steht für *unsigned*) Ganzzahlen (also **nur positive**, ganze Zahlen) und weist $2^{16}$ verschiedene Zahlen auf. Da die 0 eingeschlossen ist, ist der Wertebereich auf $0 \dots 2^{16}-1$ also $0 \dots 65535$ beschränkt. Der Elementdatentyp `int16` dagegen beinhaltet **auch negative Ganzzahlen** und enthält ebenfalls $2^{16}$ verschiedene Zahlen. Der Wertebereich beträgt dann $-\frac{2^{16}}{2} \dots \frac{2^{16}}{2}-1$ also $-32768 \dots 0 \dots 32767$.

````python
a = np.array([[-1, 0, 1, 2],
              [0, 1, 3, 1]], dtype=np.float32)  # explizite Datentyp-Zuweisung
print(a, a.dtype)
````

Der Datentyp kann mit der Methode `.astype()` geändert werden.

````python
b = a.astype(np.int8)  # Datentyp-Umwandlung
print(b, b.dtype)
````

Wird der Datentyp nicht explizit angegeben, wählt NumPy automatisch einen passenden Datentyp. Wir werden in der Regel so arbeiten und uns daher i. Allg. nicht mit dem Elementdatentyp beschäftigen. 

````python
a = np.array([-1, 0, 1, 2])  # automatische Datentyp-Zuweisung (Ganzzahl)
print(a, a.dtype)
a = np.array([0.1, 1, 10])  # automatische Datentyp-Zuweisung (Gleitkommazahl)
print(a, a.dtype)
````
