```python
import numpy as np
import matplotlib.pyplot as plt
```

## Übung 1

```python
t = np.linspace(0, 10, 20)
y_1 = 0.1 * t**2 - 0.5 * t + 0.75
y_2 = 4 * (1-np.exp(-t))

plt.plot(t, y_1, 'o--b', label=r'$y_1(t)$')
plt.plot(t, y_2, 'x:r', label=r'$y_2(t)$')
plt.xlabel('t [s]')
plt.ylabel('y [V]')
plt.title('Übung 1')
plt.grid()
plt.legend()
plt.show()
```

## Übung 2

```python
f = np.logspace(-2, 3, base=10, num=6)
A = np.linspace(-20, 80, 6)

fig, axs = plt.subplots(ncols=2, figsize=(10, 5))
fig.suptitle('Übung 2')
axs[0].plot(f, A, 's-c')
axs[0].axvspan(50, 100, color='gray')
axs[0].set_title('log. X-Skala')
axs[0].set_xscale('log')
axs[0].set_xlabel('Frequenz f [Hz]')
axs[0].set_ylabel('Amplitude A [dB]')
axs[0].grid()
axs[1].plot(f, A, 's-c')
axs[1].axvspan(50, 100, color='gray')
axs[1].set_title('lin. X-Skala')
axs[1].set_xlabel('Frequenz f [Hz]')
axs[1].set_ylabel('Amplitude A [dB]')
axs[1].grid()
plt.show()
```
