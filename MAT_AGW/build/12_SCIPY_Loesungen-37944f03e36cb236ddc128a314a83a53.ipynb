```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
```

## Übung 1
### a)

```python
data = np.loadtxt('data/Rauschsignal')
t = data[:,0]
y = data[:,1]
plt.figure(figsize=(10,5))
plt.plot(t, y, 'b')
plt.xlabel('Zeit t [s]')
plt.ylabel('Signal y [mA]')
plt.grid()
plt.show()
```

### b) 

```python
y_filt = signal.savgol_filter(y, window_length=1000, polyorder=3)
```

### c) 

```python
plt.figure(figsize=(10,5))
plt.plot(t, y, 'b', label='Rohsignal')
plt.plot(t, y_filt, 'r', lw=5, label='geglättetes Signal')
plt.xlabel('Zeit t [s]')
plt.ylabel('Signal y [mA]')
plt.legend()
plt.grid()
plt.show()
```

## Übung 2
### a)

```python
dt = 0.0001
y = np.loadtxt('data/Gausspuls.txt')
t = np.arange(0, len(y)) * dt

plt.figure(figsize=(10,5))
plt.plot(t, y, 'b')
plt.xlabel('Zeit t [s]')
plt.ylabel('Signal y [..]')
plt.grid()
plt.show()
```

### b)

```python
y_detrend = signal.detrend(y)

plt.figure(figsize=(10,5))
plt.plot(t, y, 'b', label='Originalsignal')
plt.plot(t, y_detrend, 'g', label='Trend entfernt')
plt.xlabel('Zeit t [s]')
plt.ylabel('Signal y [..]')
plt.grid()
plt.legend()
plt.show()
```

### c)

```python
q = 500  # Downsampling-Faktor
y_ds = signal.decimate(y_detrend, q)  # downgesampeltes Signal
t_ds = np.arange(0, (len(y)/q)) * dt*q  # Zeitarray für obiges Signal

plt.figure(figsize=(10,5))
plt.plot(t, y_detrend, 'b', label='originale Abtastung')
plt.plot(t_ds, y_ds, 'rx', label='Downsampling')
plt.xlabel('Zeit t [s]')
plt.ylabel('Signal y [..]')
plt.grid()
plt.legend()
plt.show()
```

## Übung 3
### a)

```python
data = np.loadtxt('data/Peaks.csv', delimiter=';', skiprows=1)
x, y = data.T

plt.figure(figsize=(10,5))
plt.plot(x, y, 'k')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()
```

### b)

```python
peaks = signal.find_peaks(y, height=0.1, distance=100)  # Peaks detektieren
peaks_inds = peaks[0]  # peaks[0] enthält die Indices der Maxima
print(x[peaks_inds])  
```

### c)

```python
plt.figure(figsize=(10,5))
plt.plot(x, y, 'k', label='Signal')
plt.plot(x[peaks_inds], y[peaks_inds], 'or', label='Maxima (Peaks)')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.legend()
plt.show()
```
