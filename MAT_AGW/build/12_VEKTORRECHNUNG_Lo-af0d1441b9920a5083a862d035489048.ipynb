```python
import numpy as np
import matplotlib.pyplot as plt
```

## Übung 1
a)

```python
a = np.array([2.1, 3.5])
b = np.array([-1.5, 4.0])
c = np.array([0.5, -0.5])

res = a + b + c
print(res)
```

b)

```python
betrag = np.linalg.norm(res) # alternativ: np.sqrt(res[0]**2 + res[1]**2)
print(betrag)
```

```python
phase = np.arctan2(res[1], res[0])  # alternativ: np.arctan(res[1]/res[0])
print(np.rad2deg(phase))  # Winkel [°]
```

c)

```python
plt.plot([0, a[0]], [0, a[1]], 'r', label=r'$\vec{a}$')
plt.plot([a[0], a[0]+b[0]], [a[1], a[1]+b[1]], 'g', label=r'$\vec{b}$')
plt.plot([a[0]+b[0], a[0]+b[0]+c[0]], [a[1]+b[1], a[1]+b[1]+c[1]], 'b', label=r'$\vec{c}$')
plt.plot([0, res[0]], [0, res[1]], 'k', label=r'$\vec{a}+\vec{b}+\vec{c}$')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.legend()
plt.show()
```

## Übung 2

a)

```python
v = np.array([1.32, 0.48])  # Geschwindigkeitsvektor [m/s]
F = np.array([81.9, 57.4])  # Kraft [N]

P = np.dot(F, v)  # Leistung via Skalarprodukt [W]
print(P)
```

b)

```python
# Der Winkel zwischen zwei Vektoren kann via Skalarprodukt berechnet werden 
# Die Definition des Skalarproduktes nach dem Winkel umstellen 
alpha = np.arccos(np.dot(F, v)/(np.linalg.norm(v)*np.linalg.norm(F)))  # Winkel [rad]
print(np.rad2deg(alpha))  # Winkel [°]
```

## Übung 3

```python
p = np.array([1, 2, 3])
phi = np.deg2rad(45)  # Drehwinkel [rad]

R = np.array([[np.cos(phi), -np.sin(phi), 0],  # Drehmatrix um z-Achse (vgl. Theorienotebook)
              [np.sin(phi), np.cos(phi),  0],
              [0,           0,            1]])

p_strich = R @ p  # gedrehter Vektor
print(p_strich)
```

## Übung 4

```python
p = np.array([5.5, 2.5])  # Vektor im xy-Koordinatensystem
t = np.array([10, 2])  # Versatz der Koordinatensysteme
phi = np.deg2rad(90)  # Verdrehung der Koordinatensysteme [rad]
R = np.array([[np.cos(phi), -np.sin(phi)],  # Drehmatrix
              [np.sin(phi), np.cos(phi)]])

p_strich = R.T @ (p - t)  # Vektor im x'y'-Koordinatensystem (Multipl. mit der transp. Drehmatrix)
print(p_strich)
```
