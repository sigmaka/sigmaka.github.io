## 🛠 Voraussetzungen

Sie sollten wissen, was ein Vektor ist und dass ein Vektor grundsätzlich auf zwei verschiedene Arten repräsentiert werden kann:

- Betrag $\vert\vec{a}\vert$ und Richtung (Winkel) $\varphi$
- Komponenten in einem Koordinatensystem (als Zeilenvektor oder Spaltenvektor)

$$\vec{a}=\left[\begin{array}{c}a_x \quad a_y \end{array}\right] \hspace{5cm} \vec{a}=\left[\begin{array}{c}a_x \\ a_y \end{array}\right]$$

In der Physik sind Vektoren mit zwei (2D) oder drei Komponenten (3D) weit verbreitet, doch auch höherdimensionale Vektoren sind grundsätzlich möglich.

## 1. Motivation

Viele Grössen in der Physik/Technik können nicht durch eine einzige Zahl eindeutig/vollständig beschrieben werden, z. B.:

- Position im Raum
- Kraft
- Geschwindigkeit, Beschleunigung
- Drehmoment
- Wärmestrom
- ...

Diese Grössen werden durch Betrag & Richtung bzw. in Komponentendarstellung (relativ zu einem Koordinatensystem) definiert und als **Vektoren** bezeichnet.

Das Gegenstück, also eine physikalische Grösse, die mit einer einzigen Zahl vollständig beschrieben werden kann, wird  **Skalar** genannt. Skalare Grössen sind z. B.:

- Temperatur
- Energie
- Leistung
- ...

## 2. Vektor

Wir definieren einen Vektor (= 1D-Array) in NumPy, indem wir dessen Komponenten definieren. Anders als in der Mathematik gibt es für einen 1D-Array in NumPy keine Orientierung (d. h. ein 1D-Array kann als Zeilen- oder Spaltenvektor aufgefasst werden). 

$$\vec{a}=\left[\begin{array}{c}a_x \\ a_y \end{array}\right]$$

<center>
    <img src="img/vector.svg" width=50% /><br>
    <small><em>Abb.: 2D-Vektor mit x- und y-Komponente</em></small>
</center><br>

☝🏼 Wir arbeiten üblicherweise im bekannten {xy}-Koordinatensystem (für 2D-Vektoren) oder im {xyz}-Koordinatensystem (für 3D-Vektoren). Zur besseren Veranschaulichung werden in den hier verwendeten Beispielen/Abbildungen immer 2D-Vektoren verwendet. Im dreidimensionalen Raum funktioniert es aber genau gleich.

```python
import numpy as np
```

```python
a = np.array([4, 2])  # 2D-Vektor mit x- und y-Komponente
print(a)
```

Der Betrag $|\vec a|$ des Vektors entspricht bekanntlich der "Länge" des Vektors und kann mit dem *Satz von Pythagoras* berechnet werden (alternativ kann die Funktion `np.linalg.norm()` verwendet werden).

```python
np.sqrt(a[0]**2 + a[1]**2)  # Betrag
# np.linalg.norm(a)  # alternativ mit NumPy
```

Die Richtung des Vektors wird mit dem Winkel $\varphi$ angegeben, der immer von der x-Achse aus gemessen wird (die positive Drehrichtung entspricht dem Gegenuhrzeigersinn). Für die Berechnung des Winkels aus den Komponenten wird bekanntlich die Arcustangensfunktion (`np.arctan()`) verwendet.

```python
np.degrees(np.arctan(a[1]/a[0]))  # Winkel in Grad [°]
```

In den Quadranten II und III ist die obige Variante nicht ideal (Vorzeichen!), darum existiert die zusätzliche Funktion `np.arctan2()`, die den Winkel in allen Quadranten korrekt zurückgibt. Dabei werden die Komponenten separat übergeben (Reihenfolge beachten!).

```python
np.degrees(np.arctan2(a[1], a[0]))  # Winkel in Grad [°]
```

### 2.1 Vektoraddition & -subtraktion

Vektoren werden komponentenweise addiert bzw subtrahiert. Grafisch entspricht dies der Aneinanderhängung der einzelnen Vektoren (man spricht in diesem Zusammenhang auch von der *Resultierenden*).

$$\vec{c}=\vec{a}+\vec{b}=\left[\begin{array}{c}a_x + b_x \\ a_y + b_y\end{array}\right] \hspace{5cm}
\vec{d}=\vec{a}-\vec{b}=\left[\begin{array}{c}a_x - b_x \\ a_y - b_y\end{array}\right]$$

<center>
    <img src="img/vector_add.svg" width=40% />&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<img src="img/vector_subtract.svg" width=40% /><br>
    <small><em>Abb.: Vektoraddition bzw. -subtraktion</em></small>
</center><br>

```python
a = np.array([4, 2])
b = np.array([1, 3])
c = a + b  # Vektoraddition
print(c)
```

```python
d = a - b  # Vektorsubtraktion
print(d)
```

### 2.2 Multiplikation von Vektoren

Bei der Multiplikation von Vektoren muss unterschieden werden zwischen:

- der komponentenweisen Multiplikation
- dem **Skalarprodukt** von zwei Vektoren
- dem **Vektorprodukt** von zwei Vektoren

#### 2.2.1 Komponentenweise Multiplikation

Die komponentenweise Multiplikation kommt bei der Multiplikation eines Vektors mit einem Skalar zum Einsatz und entspricht einer Streckung/Stauchung des Vektors. Die komponentenweise Multiplikation wird in Python mit dem normalen `*`-Zeichen realisiert.

$$t \cdot \vec{a} = \left[\begin{array}{c}t \cdot a_x \\ t \cdot a_y\end{array}\right]$$

```python
t = 2
print(t * a)  # Streckung (Verlängerung) des Vektor durch Multiplikation mit Skalar
```

Auch zwei Vektoren können komponentenweise miteinander multipliziert werden. Dies hat in der Mathematik/Physik eine untergeordnete Bedeutung, doch z. B. in der Signalverarbeitung wird dies häufig eingesetzt. Besondere Vorsicht ist geboten, wenn die beiden Arrays unterschiedliche Dimensionen haben (siehe *Broadcasting* in der NumPy-Hilfe für weitere Informationen).

$$\left[\begin{array}{c}a_x \cdot b_x \\ a_y \cdot b_y\end{array}\right]$$

```python
print(a * b)  # komponentenweise Multiplikation von Vektoren
```

#### 2.2.2 Skalarprodukt

Von grosser Bedeutung in der Mathematik/Physik ist das Skalarprodukt, welches auch als *Punktprodukt* (*dot product*) bezeichnet wird. Das Skalarprodukt liefert (wie der Name sagt) eine skalare Grösse zurück. Das Skalarprodukt "misst" sozusagen, wie parallel zwei Vektoren sind und ist gross, wenn die beiden Vektoren parallel sind bzw. 0, wenn sie senkrecht aufeinander stehen. Als Multiplikationszeichen wird für das Skalarprodukt in der Mathematik der Punkt $\bullet$ verwendet, daher der Name *Punktprodukt*.

☝🏼 Hier wird bewusst ein extragrosser Punkt verwendet (normalerweise wird ein Punkt normaler Grösse verwendet).

$$\vec{a} \bullet \vec{b} = \vert \vec{a} \vert \cdot \vert \vec{b} \vert \cdot \cos(\varphi) = a_x b_x + a_y b_y$$

<center>
    <img src="img/vector_dot.svg" width=50% /><br>
    <small><em>Abb.: Skalarprodukt</em></small>
</center><br>

Wird die obige Beziehung nach $\varphi$ umgestellt, erkennt man, dass das Skalarprodukt sehr nützlich ist, um den Winkel zwischen zwei beliebigen Vektoren zu bestimmen.

$$\varphi = \cos^{-1}\left(\frac{\vec{a} \bullet \vec{b}}{\vert \vec{a} \vert \cdot \vert \vec{b} \vert}\right)$$

Daneben hat das Skalarprodukt viele Anwendungen in der Physik/Technik.

- *Arbeit = Kraft mal Weg* $W=\vec{F} \bullet \vec{s}$
- *Leistung = Kraft mal Geschwindigkeit* $P=\vec{F} \bullet \vec{v}$
- *Leistung = Drehmoment mal Winkelgeschwindigkeit* $P=\vec{M} \bullet \vec{\omega}$

```python
skalarprodukt = np.dot(a, b)  # Skalarprodukt, alternativ: a.dot(b) oder a @ b
print(skalarprodukt)
```

#### 2.2.3 Vektorprodukt

Ebenfalls bedeutend ist das Vektorprodukt, welches auch als *Kreuzprodukt* (*cross product*) bezeichnet wird. Im Gegensatz zum Skalarprodukt wird beim Vektorprodukt ein Vektor zurückgegeben. Als Multiplikationszeichen wird für das Kreuzprodukt in der Mathematik das $\times$-Symbol verwendet, daher der Name *Kreuzprodukt*.

$$\vec{a}\times\vec{b}=\left[\begin{array}{c}a_y b_z - a_z b_y \\ a_z b_x - a_x b_z \\ a_x b_y - a_y b_x \end{array}\right]$$

<center>
    <img src="img/vector_cross.svg" width=60% /><br>
    <small><em>Abb.: Kreuzprodukt</em></small>
</center><br>

Das Vektorprodukt ist u. a. nützlich, um Flächen zu berechnen, die von Vektoren aufgespannt werden. Das Kreuzprodukt $\vec{a}\times\vec{b}$ liefert einen Vektor zurück, der senkrecht auf $\vec{a}$ und $\vec{b}$ steht und dessen Betrag der aufgespannten Fläche (Parallelogramm) von $\vec{a}$ und $\vec{b}$ entspricht. Die Richtung des Vektors kann mit der Rechten-Hand-Regel (RHR) bestimmt werden.

<center>
    <img src="img/RHR.png" /><br>
    <small><em>Abb.: Rechte-Hand-Regel (RHR) <a href="https://www.cosmos-indirekt.de/Physik-Schule/Drei-Finger-Regel">[Quelle]</a></em></small>
</center><br>

Das Kreuzprodukt hat ebenfalls viele Anwendungen in der Physik/Technik:

- *Drehmoment = Hebelarm mal Kraft* $\vec{M}=\vec{r}\times\vec{F}$
- *Bahngeschwindigkeit = Winkelgeschwindigkeit mal Radius* $\vec{v}=\vec{\omega}\times\vec{r}$

```python
a = np.array([4, 2, 0])
b = np.array([1, 3, 0])
kreuzprodukt = np.cross(a, b)  # Kreuzprodukt
print(kreuzprodukt)
```

☝🏼 Das Kreuzprodukt ist nur für dreidimensionale Vektoren definiert (die Abbildung oben verdeutlicht dies gut).

## 3. Matrix

```python
A = np.array([[1, 2, 3], [4, 5, 6]])  # Matrix (2D-Array) mit 2 Zeilen und 3 Spalten
print(A)
```

Im Gegensatz zum Vektor hat eine Matrix nun eine Orientierung, d. h. eine gewisse Anzahl Zeilen & Spalten. Wir sprechen von einer $N \times M$-Matrix mit $N$ Zeilen und $M$ Spalten. Wir kennen die Methode `.shape` bereits zur Bestimmung der Form/Gestalt eines Arrays (Anzahl Zeilen & Spalten). 

```python
A.shape
```

Die obige Matrix $A$ ist also eine 2x3-Matrix. Spiegelt man die Elemente einer Matrix an der Hauptdiagonalen (Diagonale von oben links im 45°-Winkel), so resultiert die **Transponierte** $A^\top$ der Matrix. Die Transponierte von $A$ ist nun eine 3x2-Matrix (hat also 3 Zeilen und 2 Spalten).

```python
print(A.T) # alternativ: A.transpose() oder np.transpose(A)
```

### 3.1 Rotationsmatrizen (Drehmatrizen)

Eine sehr verbreitete Anwendung von Matrizen sind Transformationen, insbesondere Rotationen, die sehr elegant mit Rotationsmatrizen realisiert werden können. Typische Anwendungsfälle sind:

- z. B. die Drehung eines Vektors um eine bestimmte Achse.
- z. B. die Transformation eines Vektors von einem Koordinatensystem in ein anderes (gedrehtes) Koordinatensystem.

#### 3.1.1 Drehung in 2D

Im zweidimensionalen Raum gibt es nur eine Achse, um die gedreht werden kann, d. h. es gibt nur eine Rotationsmatrix. Die Rotationsmatrix kann ganz einfach hergeleitet werden, indem die Drehung der Basisvektoren betrachtet wird (siehe [Anhang](#A.-Anhang)).

$$R(\varphi)=\left[\begin{array}{cc}\cos(\varphi) & -\sin(\varphi) \\ \sin(\varphi) & \cos(\varphi)\end{array}\right]$$

☝🏼 Ein zweidimensionaler Vektor weist eine grosse Ähnlichkeit mit einer komplexen Zahl $a + jb$ auf. Stellt man eine komplexe Zahl in der komplexen Ebene dar, so erkennt man dies sofort. (Das Analoge zur Drehung eines Vektor ist bei den komplexen Zahlen die Multiplikation mit $e^{j\varphi}$).

#### 3.1.2 Drehung in 3D

Im dreidimensionalen Raum gibt es nun mehrere Rotationsmatrizen, da nun um beliebig viele Achsen gedreht werden kann. Üblicherweise ist das Koordinatensystem so ausgerichtet, dass die Drehung um eine der drei Achsen (x,y,z) erfolgt.

Drehung um z-Achse: 

$$R_z(\varphi)=\left[\begin{array}{ccc}\cos(\varphi) & -\sin(\varphi) & 0 \\ \sin(\varphi) & \cos(\varphi) & 0 \\ 0 & 0 & 1\end{array}\right]$$

Drehung um y-Achse: 

$$R_y(\varphi)=\left[\begin{array}{ccc}\cos(\varphi) & 0 & \sin(\varphi) \\ 0 & 1 & 0 \\ -\sin(\varphi) & 0 & \cos(\varphi) \end{array}\right]$$

Drehung um x-Achse: 

$$R_x(\varphi)=\left[\begin{array}{ccc}1 & 0 & 0 \\ 0 & \cos(\varphi) & -\sin(\varphi) \\ 0 & \sin(\varphi) & \cos(\varphi)\end{array}\right]$$

### 3.2 Drehung eines Vektors

Ein Vektor kann nun um die entsprechende Achse gedreht werden, indem die entsprechende Rotationsmatrix mit dem Vektor multipliziert wird (Matrixmultiplikation).

$$\vec{p'}=R(\varphi)\cdot\vec{p}$$

<center>
    <img src="img/vector_Rot_active.svg" width=40% /><br>
    <small><em>Abb.: Drehung eines Vektors</em></small>
</center><br>

Die Matrixmultiplikation wollen wir an dieser Stelle nicht im Detail anschauen. Wichtig zu wissen ist, dass im Bsp. oben wieder ein Vektor zurückgegeben wird, der nun die "gedrehten" Koordinaten aufweist. Die Elemente im resultierenden Vektor entstehen durch das Skalarprodukt der Matrixzeile mit dem Originalvektor (vgl. nachfolgende Abb.), d. h. die Matrixmultiplikation basiert auf dem Skalarprodukt. 

<center>
    <img src="img/matrix_Multiplication.svg" width=60% /><br>
    <small><em>Abb.: Matrixmultiplikation</em></small>
</center><br>

Als Operator in Python wird für die Matrixmultiplikation das @-Zeichen verwendet (oder alternativ die Funktion `np.matmul()`).

```python
p = np.array([2, 2])

phi = np.deg2rad(20)  # 20° in Bogenmass umrechnen
R = np.array([[np.cos(phi), -np.sin(phi)],  # Drehmatrix
              [np.sin(phi), np.cos(phi)]])

p_strich = R @ p  # Matrixmultiplikation, alternativ: np.matmul(R, p)
print(p_strich)  # gedrehter Vektor
```

### 3.3 Koordinatentransformation

Möchte man den Vektor $\vec{p}$, der im Koordinatensystem {xy} definiert ist, in einem **gedrehten Koordinatensystem** {x'y'} ausdrücken, so lässt sich dies ebenfalls mit der Drehmatrix erreichen. Dabei wird der Vektor allerdings nun mit der Inversen der Drehmatrix multipliziert.

$$\vec{p'}=R(\varphi)^{-1}\cdot\vec{p}$$

☝🏼 Aufgrund der speziellen Form der Rotationsmatrix entspricht die Inverse einerseits der Transponierten der Rotationsmatrix und andererseits der Rotationsmatrix mit dem negativen Drehwinkel (also der Rückwärtsdrehung). Wir verwenden jeweils die Transponierte, die man mit `R.T` erhält.

$$R(\varphi)^{-1}=R(\varphi)^\top=R(-\varphi)$$

<center>
    <img src="img/vector_Rot_passive.svg" width=50% /><br>
    <small><em>Abb.: Vektor in gedrehtem Koordinatensystem</em></small>
</center><br>

```python
p = np.array([2, 2])  # Vektor im ursprünglichen Koordinatensystem

phi = np.deg2rad(20)  # 20° in Bogenmass umrechnen
R = np.array([[np.cos(phi), -np.sin(phi)],  # Drehmatrix
              [np.sin(phi), np.cos(phi)]])

p_strich = R.T @ p  # alternativ: np.matmul(R.T, p)
print(p_strich)  # Vektor in gedrehtem Koordinatensystem
```

### 3.4 Translation (Verschiebung)

#### 3.4.1 Verschiebung eines Vektors

Bisher haben wir reine Drehungen angeschaut. Eine Verschiebung eines Vektors ist nichts anderes als eine Addition eines Translationsvektors (Verschiebungsvektors) $\vec{t}$.

$$\vec{p'}=\vec{p} + \vec{t}$$

<center>
    <img src="img/vector_Trans_active.svg" width=50% /><br>
    <small><em>Abb.: Verschiebung eines Vektors</em></small>
</center><br>

Zu beachten ist, dass wenn ein Vektor sowohl **verschoben** als auch **gedreht** werden sollen, die Reihenfolge entscheidend ist. Wie üblich gilt auch hier die Regel "Punkt vor Strich".

- **Verschiebung** und dann **Drehung**

$$\vec{p'}=R(\varphi)\cdot(\vec{p} + \vec{t})$$

- **Drehung** und dann **Verschiebung**

$$\vec{p'}=R(\varphi)\cdot\vec{p} + \vec{t}$$

#### 3.4.2 Koordinatentransformation

Soll der Vektor $\vec{p}$ in einem **verschobenen Koordinatensystem** {x'y'} ausgedrückt werden, so ist der Translationsvektor $\vec{t}$ (vom usprünglichen zum verschobenen Koordinatensystem) vom Vektor $\vec{p}$ zu subtrahieren.

$$\vec{p'}=\vec{p}-\vec{t}$$

<center>
    <img src="img/vector_Trans_passive.svg" width=50% /><br>
    <small><em>Abb.: Verschiebung des Koordinatensystems</em></small>
</center><br>

```python
p = np.array([2, 2])  # Vektor im ursprünglichen Koordinatensystem
t = np.array([1, 0.5])  # Translationsvektor (vom urspr. zum verschobenen Koordinatensystem)

p_strich = p - t
print(p_strich)  # Vektor im verschobenen Koordinatensystem
```

Ist das neue Koordinatensystem {x'y'} gegenüber dem ursprünglichen sowohl **verdreht** als auch **verschoben**, so sind beide Operationen kombiniert durchzuführen.

<center>
    <img src="img/vector_RotTrans_passive.svg" width=50% /><br>
    <small><em>Abb.: Drehung und Verschiebung des Koordinatensystems</em></small>
</center><br>

$$\vec{p'}=R(\varphi)^\top\cdot(\vec{p}-\vec{t})$$

```python
p = np.array([2, 2])  # Vektor im ursprünglichen Koordinatensystem
t = np.array([1, 0.5])  # Translationsvektor (vom urspr. zum neuen Koordinatensystem)
phi = np.deg2rad(20)  # Drehwinkel des neuen Koordinatensystems gegenüber dem urspr. Koordinatensystem)

R = np.array([[np.cos(phi), -np.sin(phi)],  # Drehmatrix
              [np.sin(phi), np.cos(phi)]])

p_strich = R.T @ (p - t)  # alternativ: np.matmul(R.T, p-t)
print(p_strich)  # Vektor im neuen Koordinatensystem
```

## A. Anhang

> ☝🏼 Nachfolgend sind Informationen zur Herleitung der Drehmatrix aufgeführt, die allerdings nicht zum Stoffumfang des Moduls gehören und daher als optional zu betrachten sind.

### A.1 Herleitung der 2D-Drehmatrix 

Für die Herleitung der 2D-Drehmatrix $R(\varphi)$ betrachten wir die Drehung der Basisvektoren [1, 0] und [0, 1], d. h. der Einheitsvektoren in x-Richtung und y-Richtung. Wie Sie wissen, kann jeder beliebige Vektor als Kombination dieser beiden Basisvektoren ausgedrückt werden, d. h. mit der Drehung der Basisvektoren kann auch jeder andere Vektor gedreht werden.

#### Drehung des Basisvektors [1, 0]

Die folgende Abbildung zeigt die Drehung des Basisvektors [1, 0] um den Winkel $\varphi$. Man erkennt, dass die Koordinaten des gedrehten Vektors $\cos(\varphi)$ bzw. $\sin(\varphi)$ betragen.

$$\left[\begin{array}{c}\cos(\varphi) \\ \sin(\varphi)\end{array}\right]=R(\varphi) \cdot \left[\begin{array}{c}1 \\ 0\end{array}\right]$$

<center>
    <img src="img/vector_Rot_matrix10.svg" width=50% /><br>
    <small><em>Abb.: Drehung des ersten Basisvektors</em></small>
</center><br>

$R(\varphi)$ kann kein Skalar oder Vektor sein, da damit bei der Multiplikation nicht das korrekte Ergebnis resultieren würde. Dies ist nur möglich, wenn $R(\varphi)$ eine Matrix ist. Von dieser kennen wir nun die erste Spalte: 

$$R(\varphi)=\left[\begin{array}{cc}\cos(\varphi) & .. \\ \sin(\varphi) & ..\end{array}\right]$$

#### Drehung des Basisvektors [0, 1]

Die folgende Abbildung zeigt die Drehung des Basisvektors [0, 1] um den Winkel $\varphi$. Man erkennt, dass die Koordinaten des gedrehten Vektors $-\sin(\varphi)$ bzw. $\cos(\varphi)$ betragen.

$$\left[\begin{array}{c}-\sin(\varphi) \\ \cos(\varphi)\end{array}\right]=R(\varphi) \cdot \left[\begin{array}{c}0 \\ 1\end{array}\right]$$

<center>
    <img src="img/vector_Rot_matrix01.svg" width=50% /><br>
    <small><em>Abb.: Drehung des zweiten Basisvektors</em></small>
</center><br>

Damit kennen wir auch die zweite Spalte der Rotationsmatrix. Man kann sich nun davon überzeugen, dass die Multiplikation mit dieser Matrix nichts anderes ist als eine Überlagerung (Summe) der beiden obigen Resultate von der Drehung der Basisvektoren. 

$$R(\varphi)=\left[\begin{array}{cc}\cos(\varphi) & -\sin(\varphi) \\ \sin(\varphi) & \cos(\varphi)\end{array}\right]$$

Damit wird klar, wie die Elemente der Rotationsmatrix bestimmt werden können und warum die Struktur resultiert, die wir in Abschnitt 3.1 kennengelernt haben. Für einen beliebigen Vektor $\vec{p}$ mit den Komponenten $\left[\begin{array}{cc}p_x\\p_y\end{array}\right]$ resultiert die folgende Matrixmultiplikation und der gedrehte Vektor $\vec{p'}$.

$$\left[\begin{array}{cc}\cos(\varphi) & -\sin(\varphi) \\ \sin(\varphi) & \cos(\varphi)\end{array}\right] \cdot \left[\begin{array}{cc} p_x \\ p_y\end{array}\right] = \left[\begin{array}{cc} p_x \cdot \cos(\varphi) - p_y \cdot \sin(\varphi)  \\ p_x \cdot \sin(\varphi) + p_y \cdot \cos(\varphi)\end{array}\right] = \vec{p'} = \left[\begin{array}{cc} p'_x \\ p'_y\end{array}\right]$$

Die Rotationsmatrix hat zwei spezielle Eigenschaften:

- Wenn wir in die negative Richtung drehen, also ein negatives $\varphi$ einsetzen, ist dies gleichbedeutend mit der transponierten Drehmatrix des positiven Winkels.

$$R(-\varphi)=R(\varphi)^\top$$

- Die Länge (Betrag) der Vektoren ändert sich bei der Drehung nicht, d. h. die Determinante der Drehmatrix (sozusagen der "Betrag der Matrix") ist immer 1. Damit ist die Inverse der Drehmatrix identisch mit der Transponierten.

$$R(\varphi)^{-1}=R(\varphi)^\top$$
