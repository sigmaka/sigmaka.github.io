## Übung 2

```python
a = np.arange(-500, 500, 2)  # Endwert nicht inklusive bei arange(), drittes Argument ist Schrittweite

print(len(a))  # Anzahl Elemente
print(a[-1])  # letztes Element
print(a[22])  # Das erste Element hat Index 0, d.h. das 23. Element hat Index 22
print(a[-10:])  # Die letzten 10 Elemente
print(a[:50])  # Die ersten 50 Elemente
print(a[a>400])  # "Boolean Indexing"
print(a[(a>-100) & (a<0)])  # elementweise UND-Verküpfung mit &
```

## Übung 4

```python
A = 325
f = 50
phi = 45

T = 0.1
dt = 0.001

t = np.arange(0, T, dt)
u = A * np.sin(2*np.pi*f*t + np.deg2rad(phi))  # alternativ auch np.radians() für Umrechnung Grad -> Rad

import matplotlib.pyplot as plt  # Import der Grafik-Bibliothek (wäre typischerweise ganz oben im Script)
plt.figure(figsize=(10,5))
plt.plot(t, u, 'b')
plt.xlabel("t [s]")
plt.ylabel("u [V]")
plt.title("Spannungsverlauf")
plt.grid()
plt.show()
```
