# Modulübersicht

## 🎯 Zielsetzung

Das Modul **Angewandte Mathematik** verfolgt die folgenden übergeordneten Ziele:

- Analysieren und Bearbeiten von technischen/mathematischen Fragestellungen
- Importieren, Analysieren und Visualisieren von Messdaten
- Signalanalyse im Zeit- und Frequenzbereich (FFT)
- Aufbau eines "Werkzeugkastens" mit nützlichen Open-Source-Tools

Als untergeordnetes Ziel bzw. positiver Nebeneffekt kann die Vertiefung der Programmiersprache Python erwähnt werden, die uns bei allen Arbeiten begleiten wird. Wir werden vorwiegend in [JuypterLab](https://jupyter.org/) arbeiten, du kannst wahlweise aber auch [PyCharm](https://www.jetbrains.com/de-de/pycharm/) oder eine andere Entwicklungsumgebung (IDE) verwenden.

Die Python-Grundlagen bzw. das Arbeiten mit JupyterLab/PyCharm sollte dir bekannt sein und sind Voraussetzungen für das erfolgreiche Bestreiten des vorliegenden Moduls.

Der Fokus wird neben den mathematischen Inhalten auf verschiedenen Python-Zusatzpaketen liegen, die spezifisch für technisch-mathematische Fragestellungen bzw. die Visualisierung von Daten entwickelt wurden. Vor allem **NumPy** für numerische Berechnungen mit Arrays, **Matplotlib** für grafische Darstellungen aller Art und **SciPy** für spezielle math. Funktionen/Algorithmen werden uns durch das gesamte Modul begleiten.
