Wir wollen anhand eines einfachen Datensatzes einer Sprungantwort ein vollständiges und ansprechendes Diagramm mit Matplotlib erstellen, wobei der Fokus dabei ausnahmsweise nicht auf den Daten, sondern auf der Diagrammerstellung liegen soll. ### a) Importieren der Textdatei

- Importiere NumPy und Matplotlib.
- Importiere die Daten der Textdatei `SPRUNGANTWORT.TXT` im Ordner `data` und weise diese einem Array mit dem Namen`data` zu. Da das Textfile drei Signale enthält, hat dieser Array drei Spalten (ist also ein 2D-Array).
- Erstelle nun drei 1D-Arrays `t` (Zeit), `w` (Führungsgrösse), `x` (Regelgrösse), die jeweils einer Spalte von `data` entsprechen.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt('data/SPRUNGANTWORT.TXT', skiprows=1)
t = data[:,0]
w = data[:,1]
x = data[:,2]
```

### b) Erstellen eines Diagramms mit dem **Pyplot-Interface**

- Stelle die Führungssprungantwort dar, wobei das Diagramm eine Grösse von 12" x 5" aufweisen soll.
- Die Führungsgrösse soll als blaue, gestrichelte Linie und die Regelgrösse als rote, durchgezogene Linie dargestellt werden.
- Beschrifte die Achsen korrekt mit `Zeit [s]` sowie `Position [mm]`.
- Das Diagramm soll den Titel `Führungsprungantwort` haben und es soll eine Legende mit den Einträgen `Führungsgrösse w` bzw. `Regelgrösse x` enthalten.
- Aktiviere die Gitternetzlinien.
- Exportiere das Diagramm als SVG-Datei `Fuehrungssprungantwort.svg` in den Ordner `img`.

```python
plt.plot(t, w, 'b--', label='Führungsgrösse $w$')
```

### c) Erstellen eines Diagramms mit dem **Axes-Interface**

- Erstelle dasselbe Diagramm wie oben, doch verwende nun übungshalber das **Axes-Interface**.

```python
fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(t, w, 'b--', label='Führungsgrösse $w$')
ax.plot(t, x, 'r', label='Regelgrösse $x$')
ax.set_xlabel('Zeit $t$ [s]')
ax.set_ylabel('Position [mm]')
ax.legend()
ax.grid()
fig.savefig('img/Fuehrungssprungantwort.svg')
plt.show()
```
