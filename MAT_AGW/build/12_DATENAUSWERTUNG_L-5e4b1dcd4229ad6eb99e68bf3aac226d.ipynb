```python
import numpy as np
import matplotlib.pyplot as plt
```

## Übung 1

a)

```python
data = np.loadtxt('data/Spannungsmessung.txt', skiprows=1, delimiter=';')
U_1, U_2 = data.T
```

b)

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(U_1, label='Spannung 1', alpha=0.3)
ax.hist(U_2, label='Spannung 2', alpha=0.3)
ax.set_xlabel('Spannung U [V]')
ax.set_ylabel('Häufigkeit [-]')
ax.grid()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.boxplot([U_1, U_2], tick_labels=['1', '2'])
ax.set_xlabel('Messreihe')
ax.set_ylabel('Spannung U [V]')
ax.grid()
plt.show()
```

c)

```python
print('''
                                   Messreihe 1 \t Messreihe 2
Mittelwert:                        {:.2f} V \t {:.2f} V
Standardabweichung:                {:.3f} V \t {:.3f} V
Standardabw. des Mittelwertes      {:.3f} V \t {:.3f} V'''.format(U_1.mean(), 
                                                                  U_2.mean(),
                                                                  U_1.std(ddof=1), 
                                                                  U_2.std(ddof=1),
                                                                  U_1.std(ddof=1)/np.sqrt(len(U_1)),
                                                                  U_2.std(ddof=1)/np.sqrt(len(U_2))))
```

## Übung 2

```python
n_mess = [1072, 1104, 1288, 1239, 1276, 1290, 1160, 1223, 1171, 1173, 1178, 1236, 1413, 1401, 1224]
n_mess = np.array(n_mess)  # Liste in Array umwandeln
n = n_mess.mean()  # Mittelwert der Drehzahl n [rpm]
N = len(n_mess)  # Anzahl Messwerte
u = n_mess.std(ddof=1)/np.sqrt(N)  # Standardunsicherheit Drehzahl [rpm]
k = 2  # Erweiterungsfaktor (für 95%-Vertrauensniveau)
U = k * u  # erweiterte Messunsicherheit
'n = {:.0f} +/- {:.0f} rpm  (k = 2)'.format(n, U)  # Ausgabe des Ergebnisses
```

## Übung 3

```python
data = np.loadtxt('data/Laengenmessung.csv')
N = len(data)

L = data.mean()

u_1 = data.std(ddof=1) / np.sqrt(N)
u_2 = 0.001
u = np.sqrt(u_1**2 + u_2**2)

k = 2  # Erweiterungsfaktor (für 95%-Vertrauensniveau)
U = k * u  # erweiterte Messunsicherheit

'L = {:.4f} +/- {:.4f} mm  (k = 2)'.format(L, U)
```
